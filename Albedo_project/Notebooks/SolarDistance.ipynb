```python
import matplotlib.pyplot as plt
import numpy as np
from sunpy.coordinates.sun import earth_distance
from astropy.time import TimeDelta
from sunpy.time import TimeRange
```

 # 1.1 Sunpy example data of the distance from the sun to earth

```python
# create a time range for the year of 2011
range_2014 = TimeRange('2014-01-01', '2014-12-31')
range_2015 = TimeRange('2015-01-01', '2015-12-31')
range_2012 = TimeRange('2012-01-01', '2012-12-31')
range_2013 = TimeRange('2013-01-01', '2013-12-31')
range_2012to2015 = TimeRange('2012-01-01', '2015-12-31')
```

```python
# Calculate the distance to the sun for every day in the time range using the earth_distance function
days_2014 = np.arange(range_2014.start, range_2014.end, TimeDelta(24*60*60, format = "sec"))
days_2015 = np.arange(range_2015.start, range_2015.end, TimeDelta(24*60*60, format = "sec"))
days_2012 = np.arange(range_2012.start, range_2012.end, TimeDelta(24*60*60, format = "sec"))
days_2013 = np.arange(range_2013.start, range_2013.end, TimeDelta(24*60*60, format = "sec"))
days_2012to2015 = np.arange(range_2012to2015.start, range_2012to2015.end, TimeDelta(24*60*60, format = "sec"))

distances_2014 = earth_distance(days_2014)
distances_2015 = earth_distance(days_2015)
distances_2012 = earth_distance(days_2012)
distances_2013 = earth_distance(days_2013)
distances_2012to2015 = earth_distance(days_2012to2015)
```

```python
print(f"Average from 2012: {np.mean(distances_2012)}")
print(f"Average from 2013: {np.mean(distances_2013)}")
print(f"Average from 2014: {np.mean(distances_2014)}")
print(f"Average from 2015: {np.mean(distances_2015)}")
print(f"Average from 2012 to 2015: {np.mean(distances_2012to2015)}")

print(f"Median from 2012: {np.median(distances_2012)}")
print(f"Median from 2013: {np.median(distances_2013)}")
print(f"Median from 2014: {np.median(distances_2014)}")
print(f"Median from 2015: {np.median(distances_2015)}")
print(f"Median from 2012 to 2015: {np.median(distances_2012to2015)}")
```

```python
plt.plot(distances_2012-np.average(distances_2012), label = "2012")
plt.plot(distances_2013-np.average(distances_2013), label = "2013")
plt.plot(distances_2014-np.average(distances_2014), label = "2014")
plt.plot(distances_2015-np.average(distances_2015), label = "2015")
```
