```python
import sys, os

import numpy as np
# Note that the path ./Albedo_project/data_processing is now in the path
import xarray as xr
from cloudpathlib import AnyPath
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

# Local Imports
curdir = os.getcwd()
print(curdir+"../data_processing")
sys.path.insert(0, curdir+"/../data_processing")
sys.path.insert(0, curdir+"/../visualization")
import ceres_ebaf
import ceres_ebaf_plotting

fig_path = AnyPath("../../Figs")
data_folder="23Aug"
fig_folder="23Sep"
base_path = AnyPath("/Users/mawa7160/dev/data/CERES/EBAF/")
```

```python
ebaf_4_2_file_path = base_path / data_folder / "CERES_EBAF-TOA_Ed4.2_Subset_200003-202305.nc"
ebaf_4_2_full_years = xr.open_dataset(ebaf_4_2_file_path)
```

```python
run_len=0

solar_in_4_2 = ceres_ebaf.create_hemisphere_data(ebaf_4_2_full_years["solar_mon"], running_length=run_len)
shortwave_all_sky_4_2 = ceres_ebaf.create_hemisphere_data(ebaf_4_2_full_years["toa_sw_all_mon"], running_length=run_len)
shortwave_clear_sky_4_2 = ceres_ebaf.create_hemisphere_data(ebaf_4_2_full_years["toa_sw_clr_c_mon"], running_length=run_len)
cre_total_4_2 = shortwave_all_sky_4_2-shortwave_clear_sky_4_2

solar_diff_4_2 = solar_in_4_2["nh"]-solar_in_4_2["sh"]
cre_diff_4_2 = cre_total_4_2["nh"]-cre_total_4_2["sh"]
clear_sky_diff_4_2 = shortwave_clear_sky_4_2["nh"]-shortwave_clear_sky_4_2["sh"]
all_sky_diff_4_2 = shortwave_all_sky_4_2["nh"]-shortwave_all_sky_4_2["sh"]
```

```python
ceres_ebaf_plotting.plot_yearly_data_with_reg_line(shortwave_all_sky_4_2.year, shortwave_all_sky_4_2["global"], include_trends=True, label="SW All-Sky", title="Global Yearly Averaged Reflected Shortwave")
plt.savefig(fig_path / fig_folder / "All-Sky-Global-Timeseries.png")
```

```python
ceres_ebaf_plotting.plot_yearly_data_with_reg_line(shortwave_all_sky_4_2.year, shortwave_clear_sky_4_2["global"], include_trends=True, label="SW Clear-Sky", title="Global Yearly Averaged Clear Sky Shortwave", tick_color="red")
```

```python
ceres_ebaf_plotting.plot_hemisphere_and_global_by_year(shortwave_all_sky_4_2.year, shortwave_all_sky_4_2, include_trends=True, title="Yearly Averaged Reflected Shortwave - All Sky", set_x_ticks=False, title_fontsize=14, label_fontsize=12)
plt.savefig(fig_path / fig_folder / "All-Sky-Hemispheres-Timeseries.png")
```

```python
ceres_ebaf_plotting.plot_hemisphere_and_global_by_year(shortwave_clear_sky_4_2.year, shortwave_clear_sky_4_2, include_trends=True, title="Yearly Averaged Reflected Shortwave - Clear Sky", set_x_ticks=False, title_fontsize=14, label_fontsize=12)
plt.savefig(fig_path / fig_folder / "Clear-Sky-Hemispheres-Timeseries.png")
```

```python
ceres_ebaf_plotting.plot_yearly_data_with_reg_line(clear_sky_diff_4_2.year, clear_sky_diff_4_2, include_trends=True, label="Clear-Sky", title="Clear Sky Hemispheric Difference (NH-SH)", tick_color="orange")
plt.savefig(fig_path / fig_folder / "Clear-Sky-Difference-Timeseries.png")
```
